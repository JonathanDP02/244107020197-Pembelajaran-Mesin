```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
df = pd.read_csv('/content/spam.csv', encoding='latin-1')

# Membersihkan kolom jika menggunakan dataset SMS Spam Collection standar
# Umumnya kolom yang dipakai adalah kolom ke-0 (label) dan kolom ke-1 (teks)
if 'v1' in df.columns and 'v2' in df.columns:
    df = df[['v1', 'v2']]
    df.columns = ['label', 'text']
else:
    # Menyesuaikan jika nama kolom berbeda
    df = df.iloc[:, [0, 1]]
    df.columns = ['label', 'text']

print("--- 5 Data Teratas ---")
print(df.head())

# Mengubah label teks ('ham'/'spam') menjadi numerik (0 dan 1) jika diperlukan
df['label_num'] = df['label'].map({'ham': 0, 'spam': 1})

# Split data menjadi data latih (80%) dan data uji (20%)
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['label_num'], test_size=0.2, random_state=42
)
```

```python
print("\n--- MODEL 1: CountVectorizer ---")
cv = CountVectorizer(stop_words='english')
X_train_cv = cv.fit_transform(X_train)
X_test_cv = cv.transform(X_test)

nb_cv = MultinomialNB()
nb_cv.fit(X_train_cv, y_train)
y_pred_cv = nb_cv.predict(X_test_cv)

acc_cv = accuracy_score(y_test, y_pred_cv)
print(f"Akurasi (CountVectorizer): {acc_cv:.4f}")
print("\nClassification Report (CountVectorizer):")
print(classification_report(y_test, y_pred_cv))
```

```python
print("\n--- MODEL 2: TF-IDF ---")
tfidf = TfidfVectorizer(stop_words='english')
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

nb_tfidf = MultinomialNB()
nb_tfidf.fit(X_train_tfidf, y_train)
y_pred_tfidf = nb_tfidf.predict(X_test_tfidf)

acc_tfidf = accuracy_score(y_test, y_pred_tfidf)
print(f"Akurasi (TF-IDF): {acc_tfidf:.4f}")
print("\nClassification Report (TF-IDF):")
print(classification_report(y_test, y_pred_tfidf))
```

```python
print("\n==========================================")
print("KESIMPULAN PERBANDINGAN FITUR:")
print(f"1. Akurasi CountVectorizer : {acc_cv:.4f}")
print(f"2. Akurasi TF-IDF          : {acc_tfidf:.4f}")
if acc_tfidf > acc_cv:
    print("Kesimpulan: TF-IDF memberikan hasil yang lebih baik atau sama karena mampu mengurangi bobot kata yang terlalu sering muncul di semua dokumen namun kurang bermakna.")
else:
    print("Kesimpulan: CountVectorizer memberikan performa yang baik pada kasus data ini.")
```
